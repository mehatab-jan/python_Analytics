```python
n = 5
z = np.zeros((n,n),dtype='int')
# print(z)
z[n//2,:] = 1
z[:,n//2] = 1
print(z)
```

```python
n = 5
z = np.ones((n,n),dtype='int')
# print(z)
z[n//2,:] = 0
z[:,n//2] = 0
print(z)
```

Concat

```python
#concat and merging

df1 = pd.DataFrame({
    'A':['A0','A1','A2','A3' ],
    'B':['B0','B1','B2','B3'],
    'C':['C0','C1','C3','C4'],
    'D':['D0','D1','D2','D3']}, index=[0,1,2,3]
               )
df1
```

```python
df2 = pd.DataFrame({
    'A':['A4','A5','A6','A7' ],
    'B':['B4','B5','B6','B7'],
    'C':['C4','C5','C6','C7'],
    'D':['D4','D5','D6','D7']},index = [4,5,6,7])
df2
```

```python
concatenate = pd.concat([df1,df2])
concatenate
```

Merging

```python
df3 = pd.DataFrame({
    'E':['E4','E5','E6','E7' ],
    'F':['F4','F5','F6','F7'],
    'C':['C4','C5','C6','C7'],
    'G':['G4','G5','G6','G7']},index = [4,5,6,7])
df3
```

```python
df2 = pd.DataFrame({'A':['A']})
```

```python
df2
```

```python
merge_df = pd.merge(df1,df3,how='inner',on = 'C')
merge_df
```

```python
merge_df = pd.merge(df1,df3,how='outer',on = 'C')
merge_df
```
